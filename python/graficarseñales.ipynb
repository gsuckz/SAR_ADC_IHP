```python
import ltspice
import matplotlib.pyplot as plt
import numpy as np

# Leer el archivo .raw generado por ngspice
print("Leyendo archivo .raw...")
l = ltspice.Ltspice('./../xschem/simulations/out/sim_para_punto_2048.raw')
l.parse()

print(f"Archivo cargado exitosamente!")
print(f"Número de variables: {len(l.variables)}")
print(f"\nSeñales disponibles:")
for i, var in enumerate(l.variables):
    print(f"  {i}: {var}")

# Obtener el vector de tiempo
time = l.get_time()
print(f"\nPuntos de tiempo: {len(time)}")
print(f"Rango: {time[0]:.2e} s a {time[-1]:.2e} s")

# Función para graficar señales individuales
def plot_signal(signal_name, figsize=(14, 4)):
    """
    Grafica una señal individual
    """
    try:
        data = l.get_data(signal_name)
        
        fig, ax = plt.subplots(figsize=figsize)
        ax.plot(time, data, linewidth=1.5, color='#2E86AB')
        ax.set_xlabel('Tiempo (s)', fontsize=12, fontweight='bold')
        ax.set_ylabel('Voltaje (V)', fontsize=12, fontweight='bold')
        ax.set_title(f'Señal: {signal_name}', fontsize=14, fontweight='bold', pad=15)
        ax.grid(True, alpha=0.3, linestyle='--')
        
        # Estadísticas
        y_min, y_max = np.min(data), np.max(data)
        y_mean = np.mean(data)
        textstr = f'Min: {y_min:.4f} V\nMax: {y_max:.4f} V\nMedia: {y_mean:.4f} V'
        ax.text(0.02, 0.98, textstr, transform=ax.transAxes, 
                fontsize=10, verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
        
        plt.tight_layout()
        plt.show()
        print(f"✓ Graficada: {signal_name}")
    except:
        print(f"✗ Error al graficar: {signal_name}")

# Función para graficar múltiples señales juntas
def plot_signals_group(signal_names, title="Señales", figsize=(14, 8)):
    """
    Grafica múltiples señales en subplots
    """
    n_signals = len(signal_names)
    fig, axes = plt.subplots(n_signals, 1, figsize=figsize, sharex=True)
    
    if n_signals == 1:
        axes = [axes]
    
    for i, signal_name in enumerate(signal_names):
        try:
            data = l.get_data(signal_name)
            axes[i].plot(time, data, linewidth=1.5, color='#A23B72')
            axes[i].set_ylabel(signal_name, fontsize=11, fontweight='bold')
            axes[i].grid(True, alpha=0.3, linestyle='--')
            
            # Info rápida
            y_min, y_max = np.min(data), np.max(data)
            axes[i].text(0.99, 0.95, f'[{y_min:.3f}, {y_max:.3f}] V', 
                        transform=axes[i].transAxes, fontsize=9,
                        verticalalignment='top', horizontalalignment='right',
                        bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))
        except:
            axes[i].text(0.5, 0.5, f'Error: {signal_name}', 
                        transform=axes[i].transAxes, ha='center')
    
    axes[-1].set_xlabel('Tiempo (s)', fontsize=12, fontweight='bold')
    fig.suptitle(title, fontsize=16, fontweight='bold', y=0.995)
    plt.tight_layout()
    plt.show()

# ===== EJEMPLOS DE USO =====

# Ejemplo 1: Graficar señales digitales
print("\n" + "="*60)
print("SEÑALES DIGITALES (d0-d5)")
print("="*60)
digital_signals_1 = ['v(d0)', 'v(d1)', 'v(d2)', 'v(d3)', 'v(d4)', 'v(d5)']
plot_signals_group(digital_signals_1, "Señales Digitales d0-d5")

print("\n" + "="*60)
print("SEÑALES DIGITALES (d6-d11)")
print("="*60)
digital_signals_2 = ['v(d6)', 'v(d7)', 'v(d8)', 'v(d9)', 'v(d10)', 'v(d11)']
plot_signals_group(digital_signals_2, "Señales Digitales d6-d11")

# Ejemplo 2: Señales analógicas
print("\n" + "="*60)
print("SEÑALES ANALÓGICAS")
print("="*60)
analog_signals = ['v(vinp)', 'v(vinn)']
plot_signals_group(analog_signals, "Entradas Analógicas")

# Ejemplo 3: Señales de control
print("\n" + "="*60)
print("SEÑALES DE CONTROL")
print("="*60)
control_signals = ['v(eoc)', 'v(rst)', 'v(compout)', 'v(compp)', 'v(compn)', 'v(x1.comp_clk)']
plot_signals_group(control_signals, "Señales de Control y Comparador")

# Ejemplo 4: Graficar señales individuales específicas
print("\n" + "="*60)
print("SEÑALES INDIVIDUALES SELECCIONADAS")
print("="*60)
for signal in ['v(vinp)', 'v(vinn)', 'v(d0)', 'v(eoc)']:
    plot_signal(signal)

# ===== ANÁLISIS DE TODAS LAS SEÑALES =====
print("\n" + "="*60)
print("RESUMEN DE TODAS LAS SEÑALES")
print("="*60)
for var in l.variables:
    try:
        data = l.get_data(var)
        print(f"{var:20s} → Min: {np.min(data):8.4f} V  |  Max: {np.max(data):8.4f} V  |  Media: {np.mean(data):8.4f} V")
    except:
        print(f"{var:20s} → Error al leer")

print("\n" + "="*60)
print("TIPS:")
print("- Para graficar una señal: plot_signal('nombre_señal')")
print("- Para graficar un grupo: plot_signals_group(['señal1', 'señal2'])")
print("- Nombres de señales suelen ser en minúsculas: 'v(d0)', 'vinp', etc.")
print("="*60)
```
