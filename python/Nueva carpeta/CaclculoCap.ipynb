```python
# tal que (32 + x)/(31 + x) sea un número entero.

for x in range(1, 500):
    if (64 + x) % (63 + x) == 0:  # verifica divisibilidad exacta
        print("Solución encontrada:", x)

print("Búsqueda terminada.")
```

```python
# Buscamos x entero positivo (1 <= x <= 2^n)
# tal que (2^n + x) / (2^n - 1 + x) sea entero.

for n in range(2, 11):
    potencia = 2 ** n
    print(f"\nProbando n = {n} (2^{n} = {potencia})")
    encontrado = False

    for x in range(1, potencia + 1):  # límite superior = 2^n
        if (potencia + x) % (potencia - 1 + x) == 0:
            print(f"  ✅ Solución: x = {x}, valor entero = {(potencia + x)//(potencia - 1 + x)}")
            encontrado = True

    if not encontrado:
        print("  ❌ No hay soluciones para x ≤ 2^n.")

```
